```python
#--------------------------------------------------------------------------------
# P A N D A S 🐼
#--------------------------------------------------------------------------------

# Series = a pandas 1-dimensional labled array that can hold any data type think of it like a single column in a spreadsheet

import pandas as pd

data1 = [100, 200, 300]
data2 = [100.1, 200.2, 300.3]
data3 = ["a", "b", "c"]
data3 = [True, False, True]
data4 = {"day 1": 500, "day 2": 600, "day 3": 700, }

series = pd.Series(data1) #Series "S" caps
series = pd.Series(data1, index=["roomN0--", "b", "c"])
series.loc["c"] = 420
# print(series.iloc[1])
series = pd.Series(data4)
series.loc["day 3"] += 300


print(series)
```

```python

# DataFrame = a tabular data structure with rows and columns. (2-dimentional) similar to Excel spreedsheet but on steriods 

import pandas as pd

data = {"Name": ["captain america", "Iron man", "Thor"],
        "Age": [30, 35, 999]
        }
df= pd.DataFrame(data)
df= pd.DataFrame(data, index=["emp 1", "emp 2", "emp 3"])

# Add a new column
df["job"] = ["founder", "Ceo", "N/A"]

# Add a new row
new_row = pd.DataFrame([{"Name": "hulk", "Age": 50, "job": "Doctor"}], index=["emp 4"])
df = pd.concat([df, new_row])

# Add a new rows

new_rows = pd.DataFrame([{"Name": "spider man", "Age": 21, "job": "intern"}, 
                         {"Name": "Black panther", "Age": 30, "job": "king"}],
                          index=["emp 5", "emp 6"])

df=pd.concat([df, new_rows])



print(df)
```
