***SCALING***
-it is a preprocessing technique used to scale the values equally for all the columns 
-when numerical columns value are in different range(different scale) this technique is used to make same range
-only used for feature columns not for target columns

***WHY SACLING IS USED?***
_______________________
|age | salary | target|
|----|--------|-------|
|23  | 20000  |       |
|25  | 10000  |       |
|10  | 30000  |       |
|15  | 50000  |       |
|____|________|_______|
here linear regression gives more important to salary column bcz of high range 
so we use scaling to give importance to all columns

== there are scaling preprossessing technique ==

**STANDARDIZATION**
-type of scaling
-scales the values between no range
-if outliners are present

   **standard scaling**
     - we use this when we have less outlines (box plot to find this)
     -with the help of z score (formula) it scales the value
     -it uses mean

   **robust scaling**
     - we use this when we have more outlines (box plot to find this)
     -it uses meadian 


**NORMALIZATION**
-type of scaling
-scales the value between fixed range like (0-1)
-data is clean
