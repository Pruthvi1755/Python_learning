```python
# this ordinal encoder is used to change catagorical data into numerical data, which is required for machine learning models to process the data effectively. The encoder assigns a unique integer to each category in the dataset, allowing the model to understand and learn from the categorical features. This is particularly useful when dealing with features that have a natural order or hierarchy, as it preserves the ordinal relationship between categories.
```

```python
import pandas as pd

d = {'sales': [100, 200, 300, 400, 500],
     'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
     'size': ['small', 'medium', 'large', 'medium', 'small']
     }

df=pd.DataFrame(data=d)
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder= OrdinalEncoder(categories=[df['size'].unique()])
df['new_size']=encoder.fit_transform(df[['size']])
df.head()
```
