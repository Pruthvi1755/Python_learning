```python
# if you have a null value in data set this will help to clean those null values using mean and median in scikit learn
```

```python
import pandas as pd
import numpy as np
data = pd.DataFrame({'age':[20, 21, 19, 18, np.nan, 22, 23, np.nan], 'salary':[1000, 2000, 3000, 4000, 5000, np.nan, 7000, 8000]})

data
```

```python
data.isna().sum()
```

```python
# using simple imputer to fill the null values in data set with mean 
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
imputer.fit_transform(data)
```

```python
# using simple imputer to fill the null values in data set with median 
imputer = SimpleImputer(strategy='mean')
imputer.fit_transform(data)
```

```python
# using simple imputer to fill the null values in data set with most frequent value
imputer = SimpleImputer(strategy='most_frequent')
imputer.fit_transform(data)
```

```python
imputer = SimpleImputer(strategy='constant', fill_value=10)
imputer.fit_transform(data)
```

```python
names = pd.DataFrame({'name':['John', 'Mike', 'Sarah', 'Jessica', np.nan, 'David', 'Emily', np.nan]})
names
```
