```python
import re

data = "EnviR1onment"
# use of [] ->
res= re.search('Env[a-z][A-z][0-9]', data)
if res != None:
    print('Match Found')
else:
    print('Match not found')
```

```python
data1 = "Env@iR1onment"
# use of . ->
res1= re.search('Env.', data1)
if res1 != None:
    print('Match Found')
else:
    print('Match not found')
```

```python
data2='butterfly' 
data3 = 'buttttttterfly'
# use of {} ->
res2 = re.search('t{2}', data2)
res3 = re.search('t{2}', data3)
if res3 != None:
    print('Match Found')
else:
    print('Match not found')
```

```python
# use of ? ->

data = "Color"
data1='Colour'
res= re.search('Colou?r', data)
res= re.search('Colou?r', data1)

if res != None:
    print('Match Found')
else:
    print('Match not found')
```

```python
# use of +, * ->
# + used to find the character occurence for 1 or more
# * used to find 0 or more


import re
def multi_chr(msg, pattern):
    for pat in pattern:
        print("searching for pattern: ", p)
        print(re.findall(pat, msg))

m='sssddd........ssddd.....ssdddds.....sdddddd'
# p=['s']
# p=['s+']
# p=['s*']
# p=['sd*']
# p=['sd+']
p=['[sd]+']

multi_chr(m,p)
```

```python
# special characters \d+, \D+, \s+
import re
def multi_chr(msg, pattern):
    for pat in pattern:
        print("searching for pattern: ", p)
        print(re.findall(pat, msg))

m='a123 boy@$ bought*@(@) Himalayan28402!** and 00)) Went on a &&&*)@0ride'

p=['\d+']
# only for numeric 
p=['\D+']
# all character except numeric
p=['\s+']
# only for spaces
p=['\w+']
# except special character prints all alphabets and numbers
p=['\W+']
# only special characters





multi_chr(m,p)
```
