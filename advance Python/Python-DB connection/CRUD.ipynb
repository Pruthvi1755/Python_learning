```python


import pymysql

try: 
    con_obj = pymysql.connect(
        user='root',
        password='root'
        host='localhost'
    )
    
    cur_obj = con_obj.cursor()

    # * create new tabe in newly created db

    cur_obj.execute('create table student.std_info'
    '(name varchar(20), age int(10), degree varchar(20), stream varchar(20), marks int(10))')

    #* insert the values
    cur_obj.execute("insert into student.std_info (name,age,degree,stream,marks) values ('anup',20,'be',cse',99)")
    con_obj.commit()
     
    #* insert multiple values
    qry = "insert into student.std_info (name,age,degree,stream,marks) values(%s,%s,%s,%s,%s)"
    values = [
         ('pogba',20,'BE','cse',89),
         ('messi',30,'BE','Football',100),
         ('neymar',50,'BCA','Football',90)
         ]
    cur_obj.executemany(qry,values)
    con_obj.commit()

    
except Exception as e:
    print(e)
    
```

1. %s is a placeholder which accepts the value and sends to the respective columns 
2. execute() is responsible to handle single record, executemany() can handle multiple records 
3. exectemany() internally iterates through indivisual value present in a itarable type , each value sent to the respective place holder present in a query and then it sent to the respective column
4. rowcount --is used to count the total number of rows affected
