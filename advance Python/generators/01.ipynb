<!-- * Generator function -->
1. it is a special function which contian yeild
2. generater function yeilds the generator objects and it is internally a itarator obect ehich contains __Iter__() and __next__().
3. it rembers its priveous execution and starts the next execution from where it stops 

```python

# * normal function

def normal_function():
    print("executed 1st time")
    return 10
    print("executed 2st time")
    return 20
    print("executed 3st time")
    return 30
print(normal_function())
print(normal_function())
print(normal_function())
```

```python

# * generator function

def generator_function():
    print("executed 1st time")
    yield 10
    print("executed 2st time")
    yield 20
    print("executed 3st time")
    yield 30
gen = generator_function()
print(gen.__next__())
print(gen.__next__())
print(gen.__next__())
print(gen.__next__())




```

Note:
-the above function is a generator function which yields the value 
-it gives us the generator obj need to store in a variable and with the help of next method we can fetch one value at a time


```python

# *wap to display the squre of each element from a list using generator 

def squres(num):
    for i in num:
        yield i*i
n = [1,2,3,4,5]
gen = squres(n)
print(gen.__next__())
print(gen.__next__())
print(gen.__next__())
print(gen.__next__())
print(gen.__next__())


```

```python

# *wap to display the numbers from fibonacci using generator 
def fibonacci(num):
    a,b = 0,1
    while True:
        yield a
        a,b = b , a+b
    
num = [0,1]
gen = fibonacci(num)

for i in range (10):
    print(gen.__next__())






```
