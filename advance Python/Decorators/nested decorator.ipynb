```python
def decor1(param):
    def wrapper():
        print("-----start------")
        param()
        print("------end-------")
    return wrapper
def decor2(param):
    def wrapper1():
        print("----initialise------")
        param()
        print("----finalise------")
    return wrapper1

@decor1
@decor2
def myfun():
    print("logic!")

myfun()

```
