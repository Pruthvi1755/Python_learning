<!-- Decorators -->
1. it is a special function in python
2. it is a function which helps to change the funcanality of the exicting function whthout changing its implement
3. it is a callable wich accepts the callable and returns a callable

<!-- Rules -->
1. it must be a nested function
2. it should contains retyrn keyword and it should return in a function
3. need to declaree two function like decorating functtion and decorated function

```python
def mydecorator(para):
    def inner_fun():
        print("something happend before function call")
        para()
        print('something happend after function call')
    return inner_fun
@mydecorator
def myfunction():
    print('hi i am my function')

myfunction()
```
