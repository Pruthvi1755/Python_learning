```python
# create a class with add.items() display() update_items() delete_items()
# when obj is created we need to create empty dict automatically and perform all the above methods on the dict

class new_dict:
    def __init__(self):
        self.d1 = {}
    
    def add_items(self, k, v):
        if k not in self.d1:
            self.d1[k] = v
        else:
            print('key already exist')
    
    def display(self):
        print(self.d1)
    
    def update_items(self, k, v):
        if k in self.d1:
            self.d1[k] = v
        else:
            print("key not exists")
    
    def delete_items(self, k):
        if k in self.d1:
            del self.d1[k]
        else:
            print('key not exist')
obj = new_dict ()
obj.add_items("Luffy", "captain")
obj.add_items("Zoro", "Vice captain")
obj.add_items("samji", "Cook")
obj.display()
```
