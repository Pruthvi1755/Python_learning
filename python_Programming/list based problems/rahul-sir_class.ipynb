```python
def arrays(l, target):
    target=3
    first_idx=-1
    last_idx=-1
    for i in range(len(l)):
        if target == l[i]:
            if first_idx == -1:
                first_idx=i
            last_idx = i
    return first_idx, last_idx
l=[1,2,3,7,11,3,5,6,3,12,14]
target=int(input("target : "))
arrays(l, target)

```

```python
# waptf the product of elements in a list exculding the number in each iteration
# input-> l=[2,4,6,8,10]
# output-> [1920, 960, 640,480]
# 1920 = 4*6*8*10

def array(l):
    out=[]
    final=1
    for i in l:
        final*=i
    for i in range(len(l)):
        var=(final//l[i])
        out.append(var)
    return out
l=[2,4,6,8,10]
print(array(l))

# or

def array(l):
    final=1
    for i in l:
        final*=i
    return ([final//i for i in l])
l=[2,4,6,8,10]
print(array(l))
```
