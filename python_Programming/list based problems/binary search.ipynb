```python
l=[10,20,30,40,50,60,70,80]
key = int(input('key : '))
first=0
last=len(l)-1
while (first<=last):
    mid = (first+last)//2
    if key < l[mid]:
        last = mid-1
    elif key > l[mid]:
        first = mid+1
    else :
        print(f'key {key} is found at {mid}')
        break
else:
    print(f'key {key} is not found')

        
```
