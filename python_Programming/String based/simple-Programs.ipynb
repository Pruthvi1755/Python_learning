```python
# write a program to check the given strings are Anagram strings or not

# s1="sil ent"
# s2='lis ten'
# if len(s1)==len(s2):
#     if set(s1) == set(s2):
#          print("anagram")
#     else:
#      print('no')
# else:
#    print("no")

#* <!-- or  -->

def bubble(msg):
    l1=list(msg)
    for i in range(len(msg)):
        for j in range(i+1,len(msg)):
            if l1[i] > l1[j]:
                l1[i],l1[j]= l1[j],l1[i]
    return ''.join(l1)

s1="silent"
s2='lis ten'
if bubble(s1) == bubble(s2):
    print('anagram')
else:
    print(" no ")


```

```python

# *wap to check the given string can be a mirror string or not

def mirror(s1):
    mirror_chas = {'A','H','I','M','O', 'T', 'U', 'V','W', 'X', 'Y'}

    for i in s1:
        if i not in mirror_chas:
            return False
    return True
s1="AHI"
if mirror(s1):
    print("mirror")
else:
    print('no mirror')

```
