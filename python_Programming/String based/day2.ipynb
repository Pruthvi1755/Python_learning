```python
# wap to convert the given string based on below conditions
# 1. if length of string is even convert even number asky char to @ and space char to #
# 2. if length of a string is odd then convert odd number asky char value to $ and space char to ^

s = 'yo kaso wathasiva soulu socity vare vare Bankii senbonzukara kage yoshi'
s='I love Programming'
l = len(s)
res=""
if l % 2 == 0:
    for i in s:
        if ord(i)%2 == 0:
            res+="@"
        elif i == " ":
            res+="#"
        else:
            res+=i
else:
    for i in s:
        if ord(i)%2 !=0:
            res+="$"
        elif i == " ":
            res+="^"
        else:
            res+=i


print(res)



```
