```python
def fun(n):
    if n<4:
        print(n, end=" ")
        fun(n+1)
    print(n, end=' ')
fun(0)
```

```python
# factorial using recurssion and function no looping
def fact(n):
    if n==0:
        return 1
    return n*fact(n-1)
print(fact(5))
```
