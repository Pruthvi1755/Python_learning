reversing a number 

```python
n=int(input('n: '))
rev=0
while n>0:
    rem=n%10
    rev=rev*10+rem
    n//=10
print(rev)
```

```python
#count of dijits of n
n=int(input('n: '))
c=0
while n!=0:
    c+=1
    n//=10
print(c)
```

```python
#palindrome
n=int(input('n: '))
temp=n
rev=0
while n>0:
    rem=n%10
    rev=rev*10+rem
    n//=10
if rev==temp:
    print("palindrome")
else:
    print("not a palindrome")
```

```python
#hcf/gcd
n1=int(input('n1: '))
n2=int(input('n2: '))
hcf=1
for i in range (1,min(n1,n2)+1):
    if n1%1==0 and n2%i==0:
        hcf=i
print(hcf)
```
