write a program to check the given number is a armstrong number or not ?


153
1^3  5^3  3^3
1+125+27 = 153

```python
n=153
i=n
res=0
while n!=0:
  rem=n%10
  res+=rem**3
  n//=10
print(res)
if res==i:
  print("Armstrong")
else: print("Not Armstrong")
```

write a program to check the given number is a strong number or not ?
145
1! + 4! + 5!
145


```python
n=145
temp=n
sum=0
while n>0:
  res=1
  rem=n%10
  for i in range(1, rem+1):
    res*=i
  sum=sum+res
  n//=10  
print(sum)
if sum==temp:
  print("Strong")
else: print("Not Strong") 

  
```
