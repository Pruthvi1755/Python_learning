product of 1 to n

```python
def factorial(n):
 sum=1
 for i in range (1,n+1):
  sum*=i
 return sum

factorial(int(input("n: ")))
```

sum of even number from 1 to n

```python
def evensum(n):
  sum = 0
  for i in range(n+1):
    if i % 2 == 0:
      sum+=i
  return sum
evensum(10)
```
