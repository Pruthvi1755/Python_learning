```python
import pandas as pd
import numpy as np
# data = pd.DataFrame({'age':['ten', 'twenty', 'eleven', 'fifty', 'five'], 'salary':['1k', '2k', '3k','1k', '5k'],'study_hour':['one','two','one','four','five'],'target':[10,20,30,40,50]})
n_rows=1000
data = {
    'color': np.random.choice(['red', 'blue', 'green'], n_rows),
    'size': np.random.choice(['small', 'medium', 'large'], n_rows),
    'material': np.random.choice(['plastic', 'metal', 'wood'], n_rows),
    'price': np.random.randint(10,500, n_rows)
}
df=pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
# from category_encoders import TargetEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression


X=df.drop(columns='price')
y=df.price
x_train,x_test,y_train,y_test = train_test_split(X,y,train_size=0.2,random_state=42)
```
