1.What are the top 10 most frequent app categories in the dataset?

```python
df['Category'].value_counts().head(10).reset_index()
```

2.What is the average rating of apps in each category?

```python
df.groupby('Category')['Rating'].mean().sort_values(ascending=False).reset_index()
```

3.How many apps have a rating greater than 4.5?

```python
(df['Rating']>4.5).sum()
```

4.Which app category has the highest average rating?

```python
df.groupby('Category')['Rating'].mean().sort_index(ascending=False).head(1)
```

5.What is the most common app size in the dataset?

```python
def CleanSize(x):
  if 'M' in x:
    return float(x.replace('M', ''))
  elif 'Varies with device' == x:
    return None
  return None

df['Size'] = df['Size'].apply(CleanSize)
commonAppSize = df['Size'].value_counts().head(1)
commonAppSize
```

6.How many apps are Free vs Paid?

```python
df['Type'].value_counts()
```

7. ⁠What is the average price of paid apps?


```python
def CleanPrice(x):
  if '$' in x:
    return float(x.replace('$', ''))
  return None

df['Price'] = df['Price'].apply(CleanPrice)
df[df['Type'] == 'Paid']['Price'].mean()
```

8. ⁠What is the distribution of app installs (e.g., 1K, 10K, 1M+)?


```python
df[df['Installs'].isin(['1,000+', '10,000+', '1,000,000+'])]['Installs'].value_counts()
```

9. ⁠Which category has the highest number of installs overall?


```python
# df.groupby('Category')["Installs"].value_counts()
df.groupby('Category')['Installs'].value_counts().groupby(level=0).sum().sort_values(ascending=False).reset_index()
```
