```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("googleplaystore.csv")
```

```python
def clean_size(x):
    
    if 'M' in x:
        return float(x.replace('M', ''))
    
    elif 'k' in x:
        return float(x.replace('k', '')) / 1000
    
    return None

df['Size'] = df['Size'].apply(clean_size)
df['Size'] = df['Size'].astype(float)
```

2.How have total installs grown over the years ?

```python
df['Installs'] = df['Installs'].str.replace('[+,]', '', regex=True)
df['Installs'] = df['Installs'].replace('Free', '0')
df['Installs'] = df['Installs'].astype(float)

total_install_grown = df.groupby('Year')['Installs'].sum()

plt.plot(total_install_grown)
plt.title('Total Installs Growth')
plt.xlabel('Year')
plt.ylabel('Total Installs')
plt.show()
```

3.How has the number of apps released changed over time?


```python
df = df.drop_duplicates(subset='App')
apps_released = df.groupby('Year')['App'].count()
plt.plot(apps_released)
```

4.How has the average app price changed over time?

```python
df['Price'] = df['Price'].str.replace('$', '')
df['Price'] = df['Price'].str.replace('Everyone', '0')

df['Price']=df['Price'].astype(float)


avg_app_price = df.groupby('Year')['Price'].mean()
plt.plot(avg_app_price)
# plt.scatter(df['Year'], df['Price'])
```
